```python
from math import pi
import random

def circle(r):
    return pi * (r ** 2)

radius = round(random.random() * 10, 2)

print(f"원의 반지름: {radius}")
print(f"원주율 pi: {round(pi, 4)}")
print(f"반지름 {radius}인 원의 면적은 {round(circle(radius), 2)}")

```
