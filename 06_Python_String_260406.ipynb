f string

```python
string = 'ABC'
print(f"|{string:10}|")
print(f"|{string:>10}|")
print(f"|{string:^10}|")
print(f"|{string:-^10}|")
print(f"|{string:*>10}|")
```

```python
number = 1234
print(f"|{number:10}|")
print(f"|{number:<10}|")
print(f"|{number:-^10}|")
print(f"|{number:*>10}|")
print(f"|{number:0>10}|")
print(f"|{number:010}|")
```

format()메서드, %서식지정

```python
print("I like {} and {}.".format("Python", "Java"))
print("{1} comes after {0}.".format("A","B"))
print("{fruit} is red.".format(fruit="Apple"))

name = "Alcie"
age = 25
print("Name: %s, Age: %d" % (name, age))
```
