### 학습 내용 되돌아보기 2

```python
try:
    m = 5
    if not(1 <= m <= 12):
        raise Exception('1 ~ 12사이의 정수를 입력하세요:')

except Exception as e:
    print('예외 발생: {}'.format(e))

else:
    print('%d월: 잘 입력했습니다.' % m)
```

### 학습 내용 되돌아보기3

```python
def divide(x, y):
    return x / y

try:
    x = float(input('x 입력: '))
    y = int(input('y 입력 :'))
    divide(x, y)
except:
    print("0으로는 나눌수 없습니다")
else:
    print(f"결과 : {divide(x, y)}")
```

### 학습 내용 되돌아보기 4

```python
uyear = {1 : 'freshman', 2 :'sophomore', 3: 'junior', 4: 'senior'}

try:
    user_input = int(input('대학교 몇 학년이지요?'))
    if not(1 <= user_input <= 4):
        raise Exception
except Exception as e:
    print(f'예외 발생 이름 : {type(e)}')
    print('예외 발생 이유 : 1 ~ 4 정수를 입력하세요')

else:
    print(f'{user_input}학년 : {uyear[user_input]}')
finally:
    print('예외 처리가 잘되는군요!')
```
